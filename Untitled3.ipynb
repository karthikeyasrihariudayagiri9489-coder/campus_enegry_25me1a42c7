```python
# program 3: A function to calculate electricity cost
def calculate_bill(kwh, rate_per_unit):
    return kwh * rate_per_unit

todays_bill = calculate_bill(456, 8)
print("Today's estimated bill is Rs. ", todays_bill)
```

```python
import pandas as pd

# load the dataset you created
df = pd.read_csv("campus_enegry.csv")

# display the first few rows of the data
print("Here is the raw data:")
display(df.head())

# Group the data to see total usage per building
building_totals = df.groupby("Buliding")["kwh_used"].sum()

print("\nTotal usage per building:")
print(building_totals)

# Generate a statistical summary (average, min, max) of the numeric columns
print("\nStatistical Summary:")
display(df.describe())
```
